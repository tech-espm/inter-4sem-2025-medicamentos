join da base filtrada com marca_lab 

```python
import pandas as pd
import unicodedata
import csv

marca_lab = pd.read_csv("./marca_lab.csv")
filtrada = pd.read_csv("./base_filtrada_subgrupos.csv", sep=";", encoding="utf-8-sig")
```

```python


# inspecionar o problema
print(filtrada.shape)
print(filtrada.columns.tolist())
filtrada.head()

# detectar delimitador (opcional)
# with open("./base_filtrada_subgrupos.csv", encoding="utf-8-sig") as f:
#     sample = f.read(2048)
#     print("delimiter:", csv.Sniffer().sniff(sample).delimiter)

# reler com o separador correto (provavelmente vírgula)
filtrada = pd.read_csv("./base_filtrada_subgrupos.csv", encoding="utf-8-sig")  # ou sep="," explicitamente
print(filtrada.columns.tolist())

# checar tipos antes do merge
print(filtrada['marca'].dtype, marca_lab['idmarca'].dtype)
# alinhar tipos se necessário
# ex.: se marca for string com números:
# filtrada['marca'] = filtrada['marca'].astype(int)

# então faça o merge
df_join = filtrada.merge(marca_lab, left_on="marca", right_on="idmarca", how="inner")
df_join.head()
```

```python
# df_join.head()
df_join["laboratorio"].value_counts()

# número de laboratórios distintos
print("laboratorio (id) únicos:", df_join["laboratorio"].nunique())
```

```python
df_join.to_csv("base_filtrada_com_marcaLab.csv", index=False, encoding="utf-8-sig")
```
