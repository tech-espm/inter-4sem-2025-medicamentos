```python
import pandas as pd
import re
import unicodedata

df = pd.read_csv("./base_filtrada_com_marcaLab.csv")

df['Forma farmacêutica'].unique()

```

```python
def limpar_texto(t):
    if pd.isna(t):
        return ""
    t = str(t).lower().strip()
    # acentos
    t = unicodedata.normalize("NFKD", t)
    t = "".join(c for c in t if not unicodedata.combining(c))
    # facilitar o match
    t = t.replace("mg", " mg")
    t = t.replace("ml", " ml")
    t = t.replace("-", " ")
    t = " ".join(t.split())
    t = t.encode("ascii", errors="ignore").decode("utf-8")
    return t

df["forma_farmaceutica"] = df["Forma farmacêutica"].apply(limpar_texto)

df["forma_farmaceutica"].unique()
```

```python
def categorizar_forma(texto):
    if pd.isna(texto):
        return "outros"
    t = texto.lower()

    if "comprim" in t:
        return "comprimido"
    if "caps" in t:
        return "capsula"
    if "xarope" in t:
        return "xarope"
    if "solu" in t:
        return "solucao"
    if "got" in t:
        return "gotas"
    if "pomad" in t:
        return "pomada"
    if "creme" in t:
        return "creme"
    if "gel" in t:
        return "gel"
    
    return "outros"

df["forma_farmaceutica"] = df["forma_farmaceutica"].apply(categorizar_forma)
```

```python
# ...existing code...
df["dosagem_num"] = (
    df["dosagem"]
    .str.extract(r"(\d+\,?\d*\.?\d*)")[0]
    .str.replace(",", ".")  # Replace comma with dot
    .astype(float)
)

df["dosagem_num"]
```

```python
def extrair_unidade(texto):
    if pd.isna(texto):
        return None
    t = texto.lower()
    if "mg" in t:
        return "mg"
    if re.search(r"\bg\b", t):
        return "g"
    if "ml" in t:
        return "ml"
    return "outros"

df["dosagem_unidade"] = df["dosagem"].apply(extrair_unidade)
```

```python
df.head()
```

```python
df["quantidade_num"] = (
    df["quantidade"]
    .str.extract(r"(\d+)")
    .astype(float)
)

df["quantidade_num"]
```

```python
def tipo_quantidade(row):
    forma = row["forma_farmaceutica"]
    t = str(row["quantidade"]).lower()

    if forma in ["comprimido", "capsula", "tablet", "drágea", "sache"]:
        return "unidades"

    if forma in ["xarope", "solucao", "gotas", "ml"]:
        return "ml"

    if forma in ["pomada", "creme", "gel"]:
        return "g"

    return "outros"

df["quantidade_tipo"] = df.apply(tipo_quantidade, axis=1)
```

```python
# valores distintos (equivalente a SELECT DISTINCT)
df["quantidade_tipo"].drop_duplicates().reset_index(drop=True)

# se quiser ver também a contagem de cada tipo
df["quantidade_tipo"].value_counts()
```
